```python
# 1. Check Docker Version
!docker --version

# 2. Run the "Hello World" container
!docker run hello-world

# 3. List all containers (including stopped ones)
!docker ps -a
```

```python
# 4. Run Nginx in detached mode
!docker run -d -p 8080:80 --name my-nginx nginx

# 5. Check running containers
!docker ps
```

```python
# 6. Initialize Docker Swarm
# IMPORTANT: Replace <ip_address> with your PWD instance IP
!docker swarm init --advertise-addr <ip_address>

# 7. Create a service with 3 replicas
!docker service create --name my-web --publish 8080:80 --replicas 3 nginx
```

```python
# 8. List all services
!docker service ls

# 9. Check the status of the "my-web" service tasks
!docker service ps my-web
```

```python
# 10. Generate the worker join token (Run on Manager Node)
!docker swarm join-token worker

# 11. Verify all nodes are active (Run on Manager Node)
!docker node ls
```
